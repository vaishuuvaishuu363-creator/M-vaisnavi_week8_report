```python
df = df.drop_duplicates()
```

```python
# Clean categorical columns

df.columns = df.columns.str.strip().str.lower()

df = df.rename(columns={
    "race/ethnicity": "race_ethnicity",
    "parental level of education": "parental_education",
    "test preparation course": "test_prep",
    "math score": "math_score",
    "reading score": "reading_score",
    "writing score": "writing_score"
})
```

```python
df.columns
```

```python
#Standardize categorical text
categorical_cols = [
    "gender",
    "race_ethnicity",
    "parental_education",
    "lunch",
    "test_prep"
]

for col in categorical_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
df[categorical_cols].nunique()
```

```python
#Recap and score insepection
score_cols = [
    "math_score",
    "reading_score",
    "writing_score"
]

df[score_cols].head()
```

```python
df[score_cols].describe()
```

```python
df[score_cols].mean()
```

```python
df[score_cols].median()

```

```python
df[score_cols].std()
```

```python
#total score

df["total_score"] = df[score_cols].sum(axis=1)
df["total_score"]
```

```python
#Average Score
df["average_score"] = df[score_cols].mean(axis=1)
df["average_score"]
```

```python
#percentage
df["percentage"] =  (
    df["total_score"] / 300
) * 100
df["percentage"]
```

```python
#pass/fail status
df["result"] =  np.where(
    (df["math_score"] >= 40) &
    (df["reading_score"] >= 40) &
    (df["writing_score"] >= 40),
    "Pass",
    "Fail"
)
df["result"]
```

```python
df["result"].value_counts()
```

```python
#outliears detection

for col in score_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(col, "Potential outliers:", len(outliers))
```

```python
#Calculate average scores by parental education
score_cols = ["math_score", "reading_score", "writing_score"]

parent_scores = df.groupby("parental_education")[score_cols].mean()

parent_scores
```

```python
#create a grouped bar chart
plt.figure(figsize=(12, 6))

parent_scores = df.groupby("parental_education")[
    ["math_score", "reading_score", "writing_score"]
].mean().reset_index()

parent_scores_melted = parent_scores.melt(
    id_vars="parental_education",
    var_name="Subject",
    value_name="Average Score"
)

sns.barplot(
    data=parent_scores_melted,
    x="parental_education",
    y="Average Score",
    hue="Subject"
)

plt.title("Average Exam Scores by Parental Education")
plt.xlabel("Parental Education")
plt.ylabel("Average Score")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
#Average score by lunch type

lunch_avg = (
    df.groupby("lunch")[score_cols]
    .mean()
    .reset_index()
)

lunch_long = lunch_avg.melt(
    id_vars="lunch",
    value_vars=score_cols,
    var_name="Subject",
    value_name="Average Score"
)
plt.figure(figsize=(9,5))

sns.barplot(
    data=lunch_long,
    x="lunch",
    y="Average Score",
    hue="Subject"
)

plt.title("Average Exam Scores by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
#compare score by test preparation
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="test_prep",
    y="math_score"
)

plt.title("Math Exam Scores by Test Preparation Status")
plt.xlabel("Test Preparation Course")
plt.ylabel("Math Score")
plt.tight_layout()
plt.show()



```

```python
for subject in score_cols:
    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x="test_prep",
        y=subject
    )

    plt.title(f"{subject.replace('_', ' ').title()} by Test Preparation")
    plt.xlabel("Test Preparation Course")
    plt.ylabel("Score")
    plt.tight_layout()
    plt.show()
```

```python
#compare average scores numerically

prep_summary = (
    df.groupby("test_prep")[score_cols]
      .agg(["mean", "median", "std", "count"])
)

prep_summary
```

```python
#Math vs Reading

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="math_score",
    y="reading_score",
)

plt.title("Math vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.tight_layout()
plt.show()
```

```python
#Compare all subject pairs
pairs = [
    ("math_score", "reading_score"),
    ("math_score", "writing_score"),
    ("reading_score", "writing_score")
]
for x_col, y_col in pairs:
  plt.figure(figsize=(7,5))

  sns.scatterplot(data=df, x=x_col, y=y_col)

  plt.title(f"{x_col.title()} vs {y_col.title()}")
  plt.xlabel(x_col.replace("_", "").title())
  plt.ylabel(y_col.replace("_", "").title())
  plt.tight_layout()
  plt.show()
```

```python
#Calculate the correlation matrix
correlation_matrix = df[score_cols].corr()

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Student Exam Scores")
plt.tight_layout()
plt.show()
```
